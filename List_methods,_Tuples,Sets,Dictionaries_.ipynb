**Merge different lists together into a single list**

1) list1.extend(list2) : list2 gets merged into list1
2) new_list =list1 + list2  : you can also use + operator between 2 lists to merge them. But assign it to a new variable to access it

```python
emp_ids = [101,102,103,104] # admin1
emp_ids2 = [109,110,111,112] # admin2
```

```python
emp_ids.extend(emp_ids2) # doesn't return anything. It will update empids with empids2
print(emp_ids,emp_ids2,sep='\n')
```

```python
# another example
scores1 = [56,78,90,87]
scores2 = [34,23,12,34]
scores2.extend(scores1)
# scores2 list will be udpated with the scores1 content(scores2+scores1)
```

```python
print(scores1,scores2,sep='\n')
```

Whenever the lists need to be maintained intact(without any changes), go for +

When you want the lists to be updated with combined information , then you will go for extend

```python
name1 = ["Raman","Suman"]
name2 = ["Thaman","Baman"]
new_names = name1 + name2
print(new_names,name1,name2,sep='\n')
```
